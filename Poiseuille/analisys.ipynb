```python
import numpy as np
import matplotlib.pyplot as plt
import subprocess
import os
import time
```

```python
# TIME ANALYSIS

times = []
s = []

for steps in range(10, 10000, 200):
    start = time.time()
    subprocess.run(["solver.exe", "200", "50", "1e-4", "1.0", f"{steps}"])
    end = time.time()
    print("STEPS: ", steps, "TIME: ", end-start)
    times.append(end-start)
    s.append(steps)
    data = np.loadtxt("result.csv")
    
    if (steps - 10) % 2000 == 0:
        plt.figure()
        plt.imshow(data)
        plt.colorbar(label='Ux')
        plt.title(f"steps: {steps}")

os.remove("result.csv")

plt.figure()
plt.plot(s, times)
```
